```python
import os

# Define the target directory on Google Drive
output_dir = '/content/gdrive/My Drive/Colab Notebooks/'

# Ensure the directory exists (optional, prevents errors if folder is missing)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

for state in np.unique(data['State']):
    state_data = data[data['State'] == state]
    # Replace characters that might be invalid for filenames (like '/')
    clean_state_name = state.replace('/', '_').replace(' ', '_')
    filename = f"{output_dir}{clean_state_name}.csv"

    state_data.to_csv(filename, index=False)
    print(f"Saved: {filename}")
```

```python
import os
import numpy as np

output_dir = '/content/gdrive/My Drive/Colab Notebooks/'

if not os.path.exists(output_dir):
  os.makedirs(output_dir)

# Filter for Punjab
Punjab = data[data['State'] == "Punjab"]

# Loop through unique years in the Punjab subset
# We use Punjab['Year'] or Punjab.iloc[:, 1] to avoid the InvalidIndexError
for year in np.unique(Punjab['Year']):
    Punjab_year = Punjab[Punjab['Year'] == year]

    # Construct a valid filename
    filename = f"{output_dir}Punjab_{year}.csv"

    # Save the filtered data
    Punjab_year.to_csv(filename, index=False)
    print(f"Saved: {filename}")
```

```python
filter_gender = pd.DataFrame(df.groupby("Gender")["Total"].sum()).reset_index()
sns.catplot(x="Gender", y="Total", kind="bar", data=filter_gender);
```

```python
filter_state = pd.DataFrame(df.groupby(["State"])["Total"].sum()).reset_index()
sns.barplot(y = 'State', x = 'Total',data = filter_state, edgecolor = 'w')
plt.show()
```

```python
grouped_year = df.groupby(["Year","Gender"])["Total"].sum()
grouped_year = pd.DataFrame(grouped_year).reset_index()
# grouped_year
sns.lmplot(x="Year", y="Total", hue="Gender", data=grouped_year,height=8.27, aspect=11.7/8.27);
```

```python
#social Cause
filter_type_code = pd.DataFrame(df.groupby(["Type_code","Year"])["Total"].sum()).reset_index()
filter_type_code
sns.catplot(x="Type_code", y="Total",hue="Year", kind="bar", data=filter_type_code,height=8.27, aspect=11.7/8.27);
```

```python
filter_social_status = pd.DataFrame(df[df["Type_code"]=="Social_Status"].groupby(["Type","Gender"])["Total"].sum()).reset_index()
sns.catplot(x="Type", y="Total",hue="Gender", kind="bar", data=filter_social_status,height=8.27, aspect=11.7/8.27);
```
